```python
import heapq

def greedy_best_first(graph, h, start, goal):
    pq = [(h[start], start, [start])]
    visited = []

    while pq:
        heuristic, node, path = heapq.heappop(pq)

        if node in visited:
            continue

        visited.append(node)

        if node == goal:
            print("Path:", " -> ".join(path))
            return

        for neighbor in graph[node]:
            if neighbor not in visited:
                heapq.heappush(
                    pq,
                    (h[neighbor], neighbor, path + [neighbor])
                )

    print("No path found")


# Graph
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': ['G'],
    'E': ['G'],
    'F': ['G'],
    'G': []
}

# Heuristic values
h = {
    'A': 7,
    'B': 6,
    'C': 5,
    'D': 4,
    'E': 1,
    'F': 3,
    'G': 0
}

start = input("Enter starting node: ").upper()
goal = input("Enter goal node: ").upper()

greedy_best_first(graph, h, start, goal)
```
